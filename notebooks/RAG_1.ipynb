```python
from pypdf import PdfReader
import os

def extract_text_from_pdf(file_path):
    reader = PdfReader(file_path)

    text = ""

    for page in reader.pages:
        page_text = page.extract_text()

        if page_text:
            text += page_text + "\n"

    return text
```

```python
documents = []

for file in pdf_files:
    file_path = os.path.join(folder_path, file)

    text = extract_text_from_pdf(file_path)

    documents.append({
        "filename": file,
        "text": text
    })

print("Documents processed:", len(documents))
```

```python
from pypdf import PdfReader
import os

pages = []

for file in pdf_files:
    file_path = os.path.join(folder_path, file)
    reader = PdfReader(file_path)

    for page_number, page in enumerate(reader.pages, start=1):
        page_text = page.extract_text()

        if page_text and page_text.strip():
            pages.append({
                "filename": file,
                "page": page_number,
                "text": page_text
            })

print("Total pages with text:", len(pages))
```

```python
for page in pages[:3]:
    print("File:", page["filename"])
    print("Page:", page["page"])
    print("Text:", page["text"][:500])
    print("-" * 60)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
```

```python
chunks = []

for page in pages:
    page_chunks = splitter.split_text(page["text"])

    for chunk in page_chunks:
        chunks.append({
            "text": chunk,
            "filename": page["filename"],
            "page": page["page"]
        })

print("Total chunks:", len(chunks))
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
test_embedding = embedding_model.encode(
    "What is the minimum attendance requirement?"
)

print(test_embedding.shape)
```

```python
texts = [chunk["text"] for chunk in chunks]

embeddings = embedding_model.encode(
    texts,
    show_progress_bar=True
)

print("Number of embeddings:", len(embeddings))
print("Embedding dimensions:", embeddings.shape)
```

```python
!pip install -q faiss-cpu
```

```python
import faiss
import numpy as np

# Convert embeddings to the format FAISS expects
embedding_matrix = np.array(embeddings).astype("float32")

# Create a FAISS index
index = faiss.IndexFlatL2(384)

# Add all our chunk embeddings
index.add(embedding_matrix)

print("Number of vectors in FAISS:", index.ntotal)
```

```python
query = "What are the rules for using the laboratory?"

# Convert the question into an embedding
query_embedding = embedding_model.encode([query]).astype("float32")

# Find the 5 closest chunks
distances, indices = index.search(query_embedding, k=5)

print("Retrieved chunks:\n")

for rank, idx in enumerate(indices[0], start=1):
    print(f"--- Result {rank} ---")
    print("File:", chunks[idx]["filename"])
    print("Page:", chunks[idx]["page"])
    print("Distance:", distances[0][rank - 1])
    print("Text:")
    print(chunks[idx]["text"])
    print()
```

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    device_map="auto"
)
```

```python
def ask_rag(question, k=5):

    # 1. Convert question into an embedding
    query_embedding = embedding_model.encode(
        [question]
    ).astype("float32")

    # 2. Retrieve relevant chunks
    distances, indices = index.search(query_embedding, k)

    # 3. Build context
    context = ""

    for idx in indices[0]:
        chunk = chunks[idx]

        context += (
            f"\nSource: {chunk['filename']}, "
            f"Page: {chunk['page']}\n"
            f"{chunk['text']}\n"
        )

    # 4. Give the question + retrieved context to the LLM
    prompt = f"""You are a college knowledge assistant.

Answer the user's question using ONLY the information
provided in the context below.

If the context does not contain enough information to
answer the question, say:
"I couldn't find this information in the provided documents."

Context:
{context}

Question:
{question}

Answer:"""

    # 5. Generate answer
    result = generator(
        prompt,
        max_new_tokens=200,
        do_sample=False
    )

    answer = result[0]["generated_text"]

    return answer
```

```python
answer = ask_rag(
    "What information is provided about the laboratory?"
)

print(answer)
```

```python
answer = ask_rag(
    "How many semester and subject in Btech CSE?"
)

print(answer)
```
